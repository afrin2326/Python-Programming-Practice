# Numerical Missing Value Imputation By Class

```python
#import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
df=pd.read_csv(r"D:\ML\Bengaluru_House_Data.csv")
```

```python
df.isnull().sum()
```

```python
missing_value=df.isnull().sum()/df.shape[0]*100
missing_value
```

```python
missing_value_per=missing_value[missing_value>1].keys()
missing_value_per
```

```python
df2_drop_column=df.drop(columns=missing_value_per)
```

```python
df2_drop_column.shape
```

```python
df3_numerical=df2_drop_column.select_dtypes(include=['int64','float64'])
df3_numerical.shape
```
