```python
missing_value=df.isnull().sum()/df.shape[0]*100
missing_value
```

```python
missing_value_per=missing_value[missing_value>1].keys()
missing_value_per
```

```python
df2=df.drop(columns=missing_value_per)
```

```python
df2.shape
```

```python
df.shape
```

```python
df3=df2.select_dtypes(include=['int64','float64'])
```

```python
df3.head()
```

```python
plt.figure(figsize=(25,25))
sns.heatmap(df3.isnull())
```

```python
df3[df3.isnull().any(axis=1)]
```

```python
df3.isnull().sum()
```

```python
missing_num_var=[var for var in df3.columns if df3[var].isnull().sum()>0]
missing_num_var
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i,var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3[var],bins=20,kde_kws={'linewidth':5,'color':'#DC143C'})
```

```python
df4_mean= df3.fillna(df3.mean())
df4_mean.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i,var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3[var],bins=20,kde_kws={'linewidth':8,'color':'red'},label="Original")
    sns.distplot(df4_mean[var],bins=20,kde_kws={'linewidth':5,'color':'green'},label="Mean")
    plt.legend()
```

```python
df5_median=df3.fillna(df3.median())
df5_median.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i,var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3[var],bins=20,kde_kws={'linewidth':8,'color':'red'},label="Original")
    sns.distplot(df4_mean[var],bins=20,kde_kws={'linewidth':5,'color':'green'},label="Mean")
    sns.distplot(df5_median[var],bins=20,kde_kws={'linewidth':3,'color':'k'},label="Median")
    plt.legend()
```
