# Numpy Array Slicing (:)

```python
import numpy as np
```

```python
mx_1=np.arange(1,101).reshape(10,10)
print(mx_1)
```

```python
mx_1[4,3]
```

```python
mx_1[2]
```

```python
mx_1[:,9]
```

```python
mx_1[:,0:3]
```

```python
mx_1[:,0:3].ndim
```

```python
mx_1[1:4,7:9]
```

```python
mx_1[:,1:3]
```

```python
mx_1[:]
```

```python
mx_1[::]
```

```python
mx_1[:,:]
```

```python
mx_1.dtype
```

```python
mx_1.itemsize
```
