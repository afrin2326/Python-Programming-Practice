# plotting histogram

```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
a_students_age =np.random.randint(10,50,(70))
b_students_age =np.random.randint(20,60,(80))
```

```python
print(a_students_age)
print(b_students_age)
```

```python
plt.hist(a_students_age)
plt.title("A students age histograms")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.show()
```
