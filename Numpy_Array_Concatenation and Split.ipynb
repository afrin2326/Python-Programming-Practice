# Numpy Array Concatenation and Split

```python
import numpy as np
```

```python
arr_1=np.arange(17,33).reshape(4,4)
print(arr_1)
```

```python
arr_2=np.arange(1,17).reshape(4,4)
print(arr_2)
```

```python
list_1=[1,2,3,4]
list_2=[5,6,7,8]
```

```python
list_1+list_2
```

```python
arr_1+arr_2
```

```python
np.concatenate((arr_1,arr_2))
```

```python
print(arr_1)
print(arr_2)
```

```python
np.concatenate((arr_1,arr_2),axis=0)
```

```python
np.concatenate((arr_1,arr_2),axis=1)
```

```python
np.vstack((arr_2,arr_1))
```

```python
np.hstack((arr_2,arr_1))
```

```python
np.split(arr_1,2)
```

```python
np.split(arr_1,3)
```
