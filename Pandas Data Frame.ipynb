# How to create python pandas data frame

```python
import pandas as pd
```

```python
list=['a','b','c','d']
print(list)
```

```python
var1=pd.DataFrame(list)
print(var1)
```

```python
var1
```

```python
list2=[['a','b','c','d'],['e','f','g','h'],['i','j','k','l']]
print(list2)
```

```python
var2=pd.DataFrame(list2)
print(var2)
```

```python
var2
```

```python
dict1={'ID' : [0,1,2,3,4],'NUMBER' :  [89,76,99,65,-66]}
print(dict1)
```

```python
var3=pd.DataFrame(dict1)
print(var3)
```

```python
var3
```
